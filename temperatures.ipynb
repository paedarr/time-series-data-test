# Time-Series Data Testing

### Guilford CT Weather, Last Year

```python
import pandas as pd
import matplotlib.pyplot as plt

weather_data = pd.read_csv('guilford_weather_last_year.csv', parse_dates=['DATE'], index_col='DATE')

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

print(weather_data)
```

```python
weather_data.head()

temp_col = next((col for col in weather_data.columns if "TMAX" in col.upper()), None)

weather_data[temp_col].plot(title="Temperature")
```

```python

temp_col = next((col for col in weather_data.columns if "TMAX" in col.upper()), None)
temp_col_min = next((col for col in weather_data.columns if "TMIN" in col.upper()), None)

weather_data[temp_col].plot(title="Temperature")
weather_data[temp_col_min].plot(title="Temperature")
```
